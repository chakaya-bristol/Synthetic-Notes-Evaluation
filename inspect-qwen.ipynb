```python
import pandas as pd
import random
```

```python
# Load the data
qwen_df = pd.read_csv("synthetic_notes_qwen.csv")
print(f"Qwen notes: {len(qwen_df)}")

llama_df = pd.read_csv("synthetic_notes_llama.csv")
print(f"LLaMA notes: {len(llama_df)}")
```

```python
# Summary stats
qwen_df["word_count"] = qwen_df["generated_text"].apply(lambda x: len(str(x).split()))

print("\n--- Qwen Stats ---")
print(qwen_df["word_count"].describe())

llama_df["word_count"] = llama_df["generated_text"].apply(lambda x: len(str(x).split()))

print("\n--- LlaMa Stats ---")
print(llama_df["word_count"].describe())
```
